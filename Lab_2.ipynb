```python
import re
import string
```

**Experiment 2.1:**  Write a Python program to perform basic text preprocessing on a given data. The program should:
* Convert text to lowercase. (Also display how many uppercase?)
* Remove punctuation.        (Also display how many different types of punctuations?)
* Remove numbers.            (Also display how many numbers?)
* Remove extra whitespace. (How much extra whitespace is present?)
* Display the cleaned text.     


```python
file_path = "/content/drive/MyDrive/CLNLP_Lab/2.1_text_data.txt"

with open(file_path, "r", encoding="utf-8") as file:
    raw_text = file.read()

uppercase_count = sum(1 for char in raw_text if char.isupper())
text_lowercase = raw_text.lower()
```

```python
distinct_punctuating_types = set(
    char for char in raw_text if char in string.punctuation
)
num_punct_types = len(distinct_punctuating_types)
text_no_punctuation = text_lowercase.translate(
    str.maketrans("", "", string.punctuation)
)
```

```python
digit_count = sum(1 for char in raw_text if char.isdigit())
number_sequences = re.findall(r"\d+", raw_text)
text_no_numbers = re.sub(r"\d+", "", text_no_punctuation)
```

```python
whitespace_matches = re.finditer(r"\s+", raw_text)
extra_whitespace_count = sum(len(m.group(0)) - 1 for m in whitespace_matches)
cleaned_text = re.sub(r"\s+", " ", text_no_numbers).strip()
```

```python
print("=" * 50)
print("             TEXT PREPROCESSING REPORT           ")
print("=" * 50)
print(f"1. Uppercase characters count : {uppercase_count}")
print(
    f"2. Punctuation types count    : {num_punct_types} {sorted(list(distinct_punctuating_types))}"
)
print(
    f"3. Numbers/Digits count       : {digit_count} digits across {len(number_sequences)} numbers {number_sequences}"
)
print(f"4. Extra whitespace count     : {extra_whitespace_count} characters")
print("=" * 50)
print("\n--- CLEANED TEXT ---\n")
print(cleaned_text)
```

**Experiment 2.2:** Write a Python program to perform word tokenization and sentence tokenization on a given text. Display the resulting words and sentences separately.


**2.2.1. Using NLTK**


```python
import nltk
from nltk.tokenize import sent_tokenize, word_tokenize
```

```python
with open("/content/drive/MyDrive/CLNLP_Lab/2.2_tokenization_data.txt", "r", encoding="utf-8") as file:
    text_data = file.read()
```

```python
nltk.download('punkt_tab')
sentences = sent_tokenize(text_data)
```

```python
words = word_tokenize(text_data)
```

```python
print("=== SENTENCE TOKENIZATION (NLTK) ===")
for i, sentence in enumerate(sentences, 1):
    print(f"{i}. {sentence}")

print("\n=== WORD TOKENIZATION (NLTK) ===")
print(words)
```

**2.2.2. Using spaCy**

```python
import spacy
nlp = spacy.load("en_core_web_sm")
```

```python
with open("/content/drive/MyDrive/CLNLP_Lab/2.2_tokenization_data.txt", "r", encoding="utf-8") as file:
    text_data = file.read()
```

```python
doc = nlp(text_data)
```

```python
sentences = [sent.text.strip() for sent in doc.sents]
words = [token.text for token in doc]
```

```python
print("=== SENTENCE TOKENIZATION (spaCy) ===")
for i, sentence in enumerate(sentences, 1):
    print(f"{i}. {sentence}")

print("\n=== WORD TOKENIZATION (spaCy) ===")
print(words)
```

**2.2.3. Without NLTK or spaCy**

```python
with open("/content/drive/MyDrive/CLNLP_Lab/2.2_tokenization_data.txt", "r", encoding="utf-8") as file:
    text_data = file.read()
```

```python
sentences = [
    s.strip() for s in re.split(r"(?<=[.!?])\s+", text_data) if s.strip()
]
```

```python
words = re.findall(r"\b\w+\b|[^\w\s]", text_data)
```

```python
print("=== SENTENCE TOKENIZATION (Pure Python) ===")
for i, sentence in enumerate(sentences, 1):
    print(f"{i}. {sentence}")

print("\n=== WORD TOKENIZATION (Pure Python) ===")
print(words)
```

**Experiment 2.3:** Write a Python program to tokenize a text and remove commonly occurring stop words such as is, the, a, an, in, of, and, to, etc. Display the original and filtered tokens.


```python
from nltk.corpus import stopwords
from nltk.tokenize import word_tokenize
```

```python
with open("/content/drive/MyDrive/CLNLP_Lab/2.3_clean_data.txt", "r", encoding="utf-8") as file:
    text_data = file.read()
```

```python
original_tokens = re.findall(r"\b\w+\b", text_data)
```

```python
nltk.download('stopwords')
stop_words_list = set(stopwords.words("english"))
```

```python
stop_words_found = [
    token for token in original_tokens if token.lower() in stop_words_list
]
filtered_tokens = [
    token for token in original_tokens if token.lower() not in stop_words_list
]
```

```python
print("=== 1. ORIGINAL TOKENS ===")
print(original_tokens)

print("\n=== 2. STOP WORDS REMOVED (SEPARATE DISPLAY) ===")
print(stop_words_found)

print("\n=== 3. FILTERED TOKENS (WITHOUT STOP WORDS) ===")
print(filtered_tokens)

print("\n=== SUMMARY STATISTICS ===")
print(f"Total Original Tokens : {len(original_tokens)}")
print(f"Total Stop Words      : {len(stop_words_found)}")
print(f"Total Filtered Tokens  : {len(filtered_tokens)}")
```
