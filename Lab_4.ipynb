# **Experiment 4: Term Frequency and NER**

```python
import csv
import string
from collections import Counter
import spacy
import math
```

**4.1. To perform Term-Frequency Analysis and Named Entity Recognition on a given text using any toolkit.**



*   Calculate the frequency of each word and save the output in CSV format (Term, Frequency)
*   Identify the top 10 most frequent terms and display their corresponding frequencies.




```python
file_path = "/content/drive/MyDrive/CLNLP_Lab/4.1_4.2_input.txt"
with open(file_path, "r", encoding="utf-8") as file:
    text_data = file.read()
```

```python
clean_text = text_data.lower().translate(
    str.maketrans("", "", string.punctuation)
)
words = clean_text.split()

word_freq = Counter(words)
```

```python
with open("term_frequency_toolkit.csv", "w", newline="", encoding="utf-8") as f:
    writer = csv.writer(f)
    writer.writerow(["Term", "Frequency"])
    for term, freq in word_freq.items():
        writer.writerow([term, freq])

print("=== TOP 10 MOST FREQUENT TERMS ===")
for term, count in word_freq.most_common(10):
    print(f"{term:<15} : {count}")
```

```python
nlp = spacy.load("en_core_web_sm")
doc = nlp(text_data)

print("\n=== NAMED ENTITIES IDENTIFIED (NER) ===")
print(f"{'Entity':<30} | {'Label':<15} | {'Explanation'}")
print("-" * 65)
for ent in doc.ents:
    print(f"{ent.text:<30} | {ent.label_:<15} | {spacy.explain(ent.label_)}")
```

**4.2. To perform Term-Frequency Analysis on a given text without using any toolkit.**

*   Calculate the frequency of each word and save the output in CSV format (Term, Frequency)
*   Identify the top 10 most frequent terms and display their corresponding frequencies.



```python
text_lower = text_data.lower()
text_no_punct = text_lower.translate(str.maketrans("", "", string.punctuation))
words = text_no_punct.split()
```

```python
frequency_dict = {}
for word in words:
    if word in frequency_dict:
        frequency_dict[word] += 1
    else:
        frequency_dict[word] = 1
```

```python
with open("term_frequency_pure.csv", "w", newline="", encoding="utf-8") as f:
    writer = csv.writer(f)
    writer.writerow(["Term", "Frequency"])
    for term, freq in frequency_dict.items():
        writer.writerow([term, freq])
sorted_terms = sorted(frequency_dict.items(), key=lambda x: x[1], reverse=True)

print("=== TOP 10 MOST FREQUENT TERMS (Pure Python) ===")
print(f"{'Term':<15} | {'Frequency'}")
print("-" * 30)
for term, freq in sorted_terms[:10]:
    print(f"{term:<15} | {freq}")
```

**4.3. Write a Python program without using NLTK, spaCy, or any other NLP toolkit to:**


*   Read multiple text documents.



```python
documents = [
    "Natural language processing is a field of artificial intelligence.",
    "Natural language processing helps computers understand human language.",
    "Machine learning is an important part of artificial intelligence.",
]
```



*   Convert text to lowercase.
*   Remove punctuation.




```python
tokenized_docs = []
for doc in documents:
    clean_doc = doc.lower().translate(str.maketrans("", "", string.punctuation))
    tokens = clean_doc.split()
    tokenized_docs.append(tokens)
```

*   Tokenize the documents.

```python
print("=== 1. TOKENIZED DOCUMENTS ===")
for i, tokens in enumerate(tokenized_docs, 1):
    print(f"Document {i}: {tokens}")
```

*   Calculate TF for every term.
*   Calculate DF for every term.
*   Calculate IDF using the logarithmic formula.
*   Calculate TF-IDF for every term in every document.
*   Display Term, TF, DF, IDF, and TF-IDF at each step.

```python
tf_docs = []
for tokens in tokenized_docs:
    total_words = len(tokens)
    doc_tf = {}
    for token in tokens:
        doc_tf[token] = tokens.count(token) / total_words
    tf_docs.append(doc_tf)
vocab = sorted(list(set(word for doc in tokenized_docs for word in doc)))
N = len(documents)

df_dict = {}
idf_dict = {}

for term in vocab:
    doc_count = sum(1 for doc in tokenized_docs if term in doc)
    df_dict[term] = doc_count

    idf_dict[term] = math.log(N / doc_count)
print("\n=== 2. DF AND IDF VALUES FOR ALL TERMS ===")
print(f"{'Term':<15} | {'DF':<5} | {'IDF':<10}")
print("-" * 35)
for term in vocab:
    print(f"{term:<15} | {df_dict[term]:<5} | {idf_dict[term]:.4f}")
tfidf_docs = []

print("\n=== 3. STEP-BY-STEP TF, DF, IDF, AND TF-IDF PER DOCUMENT ===")
for i, (tokens, tf_map) in enumerate(zip(tokenized_docs, tf_docs), 1):
    print(f"\n--- Document {i} Calculations ---")
    print(
        f"{'Term':<15} | {'TF':<10} | {'DF':<5} | {'IDF':<10} | {'TF-IDF':<10}"
    )
    print("-" * 55)

    doc_tfidf = {}
    for term in dict.fromkeys(tokens):
        tf_val = tf_map[term]
        df_val = df_dict[term]
        idf_val = idf_dict[term]
        tfidf_val = tf_val * idf_val

        doc_tfidf[term] = tfidf_val
        print(
            f"{term:<15} | {tf_val:<10.4f} | {df_val:<5} | {idf_val:<10.4f} | {tfidf_val:<10.4f}"
        )

    tfidf_docs.append(doc_tfidf)
```


*   Identify the top 10 terms with the highest TF-IDF score for each document.

```python
print("\n=== 4. TOP TERMS WITH HIGHEST TF-IDF PER DOCUMENT ===")
for i, doc_tfidf in enumerate(tfidf_docs, 1):
    # Sort terms by TF-IDF value descending
    sorted_terms = sorted(doc_tfidf.items(), key=lambda x: x[1], reverse=True)

    print(f"\nTop Terms for Document {i}:")
    for rank, (term, score) in enumerate(sorted_terms[:10], 1):
        print(f"  {rank}. {term:<15} : {score:.4f}")
```
