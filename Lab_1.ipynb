# **Lab 1: Data Analysis**

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df = pd.read_csv("/content/drive/MyDrive/CLNLP_Lab/employee_information_100.csv")
```

**Experiment-1.1: Read the employee dataset and display the average salary of each department using a bar chart.**


```python
avg_salary = df.groupby('Department')['Salary'].mean()
```

```python
plt.figure(figsize=(8,4))
avg_salary.plot(kind='bar', color = 'lightblue', edgecolor = 'darkblue')
plt.title('Average Salary by Department')
plt.xlabel('Department')
plt.ylabel('Average Salary')
plt.show()
```

**Experiment 1.2: Display the number of employees working in each department.**

```python
emp_count = df.groupby('Department')['Employee_ID'].count()
print("Number of employees working in each department:")
print(emp_count)
```

**Experiment 1.3:Display the percentage of male and female employees using a pie chart**

```python
gender_counts = df['Gender'].value_counts()
plt.figure(figsize=(6,6))
plt.pie(gender_counts, labels=gender_counts.index, autopct='%1.1f%%', startangle=90)
plt.title('Percentage of Male and Female Employees')
plt.show()
```

**Experiment 1.4: Draw a histogram showing salary distribution.**

```python
plt.figure(figsize=(8,4))
plt.hist(df['Salary'], bins=10, color = 'lightblue', edgecolor='black')
plt.title('Salary Distribution')
plt.xlabel('Salary')
plt.ylabel('Frequency (Number of employees)')
plt.show()
```

**Experiment 1.5: Draw a scatter plot showing the relationship between experience and salary.**

```python
plt.figure(figsize=(10,6))
sns.scatterplot(data=df, x='Experience_Years', y='Salary', hue='Department', palette='viridis', edgecolor='darkblue', s=80) # s for size of markers
plt.title('Experience vs Salary by Department')
plt.xlabel('Experience')
plt.ylabel('Salary')
plt.legend(bbox_to_anchor=(1.05, 1), loc='upper left') # Place legend outside the plot
plt.tight_layout()
plt.show()
```

**Experiment 1.6: Display the top 10 employees with the highest salary.**

```python
top_10_salaries = df.nlargest(10, 'Salary')
print("Top 10 employees with the highest salary:")
print(top_10_salaries)
```

**Experiment 1.7: Find the highest salary in every department.**

```python
highest_salary_per_dept = df.groupby('Department')['Salary'].max()
print("Highest salary in every department:")
print(highest_salary_per_dept)
```

**Experiment 1.8: Display employees whose salary is greater than the overall average salary.**

```python
overall_avg_salary = df['Salary'].mean()
high_salary_employees = df[df['Salary'] > overall_avg_salary]
print("Employees with salary greater than the overall average salary:")
print(high_salary_employees)
```

**Experiment 1.9: Calculate the average years of experience in each department.**

```python
avg_exp_per_dept = df.groupby('Department')['Experience_Years'].mean()
print("Average years of experience in each department:")
print(avg_exp_per_dept)
```

**Experiment 1.10: Display the distribution of employee ages using a histogram.**

```python
plt.figure(figsize=(8,4))
plt.hist(df['Age'], bins=10, color = 'lightblue', edgecolor='darkblue')
plt.title('Employee Age Distribution')
plt.xlabel('Age')
plt.ylabel('Frequency (Number of employees)')
plt.show()
```
